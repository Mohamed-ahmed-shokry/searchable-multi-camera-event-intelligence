# Phase 10 — Searchable Outputs (`intelligence.db`) Exploration

This notebook demonstrates:
1. Ingesting upstream detections, tracks, trajectories, events, and zone/line counts into `intelligence.db`.
2. Querying structured events by `event_type`, `zone_id`, `track_id`, time range, and minimum duration.
3. Querying tracks by duration and ground-plane `path_length_px`.
4. Resolving and verifying visual evidence (`snapshot_path` and `crop_path`) on disk.

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().resolve().parents[1]
SCRIPTS_DIR = PROJECT_ROOT / "cv_projects" / "10_searchable_outputs" / "scripts"
if str(SCRIPTS_DIR) not in sys.path:
    sys.path.insert(0, str(SCRIPTS_DIR))

from search_events import (
    load_search_config,
    lookup_event_evidence,
    query_events,
    query_tracks,
    run_searchable_outputs,
)
```

```python
config_path = (
    PROJECT_ROOT / "cv_projects" / "10_searchable_outputs" / "configs" / "search.yaml"
)
output_dir = (
    PROJECT_ROOT
    / "cv_projects"
    / "10_searchable_outputs"
    / "outputs"
    / "people_detection"
)
config = load_search_config(config_path)
summary = run_searchable_outputs(config, output_dir, project_root=PROJECT_ROOT)
summary.to_dict()
```

```python
db_path = output_dir / "intelligence.db"
dwell_events = query_events(
    db_path, event_type="dwell_time_event", zone_id="right_aisle"
)
for ev in dwell_events:
    print(ev.to_dict())
```

```python
long_tracks = query_tracks(db_path, min_duration=5.0)
for trk in long_tracks:
    print(trk.to_dict())

evidence = lookup_event_evidence(db_path, "evt_0002")
print(evidence.to_dict())
```
