# Phase 3 — Person Tracking: ByteTrack vs BoT-SORT Comparison

This notebook compares ByteTrack (`bytetrack.yaml`) and BoT-SORT (`botsort.yaml`) on MOT17-11-FRCNN and static surveillance clips, inspecting MOTA, IDF1, ID switches, and track duration distributions.


```python
import json
import sys
from pathlib import Path

import cv2
import matplotlib.pyplot as plt

SCRIPTS_DIR = Path.cwd().parent / "scripts"
sys.path.insert(0, str(SCRIPTS_DIR))
INSPECTION_DIR = Path.cwd().parents[1] / "01_video_inspection" / "scripts"
sys.path.insert(0, str(INSPECTION_DIR))
```

## 1. Run Side-by-Side Tracker Checkpoint on MOT17-11-FRCNN


```python
from compare_trackers import compare_trackers
from track_people import draw_tracks, read_tracks_csv
from video_source import VideoSource

mot_seq = Path.cwd().parents[2] / "data" / "mot17" / "MOT17-11-FRCNN"
checkpoint = compare_trackers(
    mot_seq,
    output_dir=Path.cwd().parent / "outputs" / "notebook_mot17_cmp",
    max_frames=150,
)
print(json.dumps(checkpoint["trackers"], indent=2))
```

## 2. Visualize Annotated Tracking Frame


```python
tracks_csv = checkpoint["trackers"]["bytetrack"]["tracks_csv"]
frame_tracks = read_tracks_csv(tracks_csv)
with VideoSource(mot_seq) as src:
    sample_frame = src.read_frame(50)

rendered = draw_tracks(sample_frame, frame_tracks[50].tracks)
plt.figure(figsize=(12, 7))
plt.imshow(cv2.cvtColor(rendered, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.title("ByteTrack on MOT17-11-FRCNN (Frame 50)")
plt.show()
```
